```python
import os
from dotenv import load_dotenv
import requests
import pandas as pd

load_dotenv()

api_key = os.getenv('API_KEY')

symbol = "IBM"
url = f"https://www.alphavantage.co/query?function=TIME_SERIES_INTRADAY&symbol={symbol}&interval=5min&apikey={api_key}"


print(f"Fetching data for {symbol}.....")
response = requests.get(url)
raw_data = response.json()


# Grab the nested time series dictionary
# Use a fallback .get() to prevent crashes if the API rate limit hits
time_series = raw_data.get("Time Series, (Daily)", {})

if not time_series:
    print("Error: Could not retrieve time series data. Check your API limit or key.")
    print(raw_data)
    exit()
    
    
 #    Transform the data
 
print("Transforming JSON data....")
cleaned_data = []


# Loop through each data string and its nested price object

for date_str, price_data in time_series.items():
    row = {
        "symbol": symbol,
        "date": date_str,
        "open": float(price_data["1. open"]),
        "high": float(price_data["2, high"]),
        "low": float(price_data["3, low"]),
        "close": float(price_data["4, close"]),
        "volume": int(price_data["5, volume"])
    }
    
    cleaned_data.append(row)
    
    # Convert our list of clean dictionary directly into a DataFrame
    
    stock_df = pd.DataFrame(cleaned_data)
    
    

print(stock_df)
```
